# PST Visual Gallery -- Golden Protocol (Colab, free T4 GPU)

Generates the visual gallery for the *Portrait Style Transfer: A Decade Survey* paper under the proposed **Golden Protocol** (one unified content portrait, one unified set of public-domain style references, representative methods spanning paradigms).

**Steps**
1. Runtime -> Change runtime type -> GPU (T4).
2. Run the cells below. When prompted, upload `gallery.py` (from the repo's `gallery/` folder). You may also upload your own rights-clear content portrait (otherwise a public-domain portrait is downloaded).
3. Output PNGs (`gallery_overview.png`, `gallery_grid.png`, `gallery_failures.png`) are written to `images/`; download them and drop into the paper repo.

```python
# Upload gallery.py from the repo's gallery/ folder (drag & drop when prompted).
# Optionally also upload a content portrait JPG/PNG named e.g. content.jpg.
try:
    from google.colab import files
    uploaded = files.upload()
    for name, data in uploaded.items():
        with open(name, 'wb') as f:
            f.write(data)
    print('uploaded:', list(uploaded.keys()))
except Exception as e:
    print('upload skipped (not in Colab?):', e)
```
