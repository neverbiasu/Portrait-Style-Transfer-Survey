# PST Visual Gallery -- Golden Protocol (Colab, free T4 GPU)

Generates the visual gallery for the *Portrait Style Transfer: A Decade Survey* paper under the proposed **Golden Protocol**: six attribute-matched (content, style) pairs, eight representative methods spanning four families.

**Steps**
1. Runtime -> Change runtime type -> GPU (T4).
2. Run the cells below. They clone this survey repo, install deps, and run the engine. Output PNGs are written to `images/`.
3. Download `gallery_overview.png`, `gallery_grid.png`, `gallery_failures.png` and drop them into the paper repo.

Each method self-provisions its code + weights (weights download from Google Drive / HuggingFace on Colab) and runs in an isolated batch; a failure in one method keeps the rest.
